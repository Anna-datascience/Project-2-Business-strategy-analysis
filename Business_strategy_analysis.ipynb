```python
def tmp_error_print(msg):
    print(f"\033[91mERROR: {msg}\033[0m")

def dropna_by_column(dataframe: pd.DataFrame, column: str) -> pd.DataFrame:
    '''

    '''
    bef = len(dataframe)
    returndf = dataframe.dropna(axis=0, subset=[column])
    aft = len(returndf)
    print(f"{bef - aft} rows droped in dataframe") #bef - aft because if other way round number is negative
    return returndf

def try_to_numeric(dataframe: pd.DataFrame, column: str, name: str = "dataframe", return_faulty_ones: bool = False) -> pd.DataFrame:
    #TODO: differentiate try_to_numeric and return_not_numeric_lines (get_non_numeric_rows does it,
    #TODO: remove return_faulty_values-functionality)
    """
    Tries casting values of column from dataframe to numeric.

    Args:
        dataframe (DataFrame): The data to use.
        column (str): The column name for wich data to cast.
        name (str): Optional parameter to name the data.
        return_faulty_ones (bool):

    Returns:
        pd.DataFrame: If it worked, dataframe with column casted to numeric,
        if it didnt work and return_faulty_ones is False it returns dataframe unchanged
        if it didnt work and return_faulty_ones is True it returns a dataframe with the rows that arent castable.
    """
    worked = True
    returnvalue = dataframe
    try:
        returnvalue[column] = pd.to_numeric(dataframe[column])
    except ValueError:
        worked = False
    finally:
        if worked:
            print(f"Column {column} in {name} is now numeric.")
        else:
            tmp_error_print(f"Failure in casting column {column} in {name} to numeric!")
    if return_faulty_ones and not worked:
        print(f"Return rows with faulty values of {name} in column {column}")
        mask = ((pd.to_numeric(dataframe[column], errors="coerce").isna()) & (dataframe[column].notna()))
        returnvalue = dataframe.loc[mask, :]
        #returnvalue = dataframe.loc[~dataframe[column].str.isnumeric(), :] #-> WRONG! isnumeric only looks at single signs. Everything that is not 0..9 is not numeric
    return returnvalue


def get_non_numeric_rows(dataframe: pd.DataFrame, column: str, name: str = "dataframe") -> pd.DataFrame:
    mask = ((pd.to_numeric(dataframe[column], errors="coerce").isna()) & (dataframe[column].notna()))
    notnumeric = dataframe.loc[mask, :]
    total_rows = dataframe.shape[0]
    percent_missing = (100*len(notnumeric)/total_rows)
    print(f"Amount of rows not numeric: {len(notnumeric)}. Represents {percent_missing}% in {name}")
    #return notnumeric
```

### 4. What are the most sold products?

```python
# adding product name to orderlines table to identify product (merge via left join)

orderlines_product_name = orderlines_qu.merge(
    products_df[['sku', 'name']],
    how='left',
    on='sku'
)

# sorting by top 10 sku's with highest product quantity sold

most_sold = (orderlines_product_name.groupby(
    'sku').agg({'product_quantity': 'sum', 'name': 'first'})).sort_values(by='product_quantity', ascending=False).head(10)

most_sold
```

### 5. What are the products that generate the most revenue?

```python
# sorting by top 10 sku's with highest unit_price_total

most_revenue = (orderlines_product_name.groupby(
    'sku').agg({'unit_price_total': 'sum', 'name': 'first'})).sort_values(by='unit_price_total', ascending=False).head(10)

most_revenue
```

### 6. How many products are being discounted?

```python
# merging orderlines & products

discounts = orderlines_qu.merge(
    products_df,
    how='left',
    on='sku'
)

discounts = discounts.drop(['id','unit_price_total'], axis=1)

discounts
```

```python
# create helper column with first 3 letters of SKU

discounts['short'] = discounts['sku'].str[:3]


# merge 'brands' columns into new table
discounts = discounts.merge(brands_df,
                how="left",
                on='short')

# drop unnecessary columns
#discounts.drop(columns = ['long_x', 'long_y'], inplace=True)


# convert 'type' from float to integer
discounts['type'] = discounts['type'].astype('Int64')

discounts.head()
```

```python
#create 'discount' column

discounts['discount'] = discounts['price'] - discounts['unit_price']


# reorder columns for clearer view
discounts = discounts[
    ['id_order', 'sku', 'date', 'product_quantity', 'unit_price', 'price',
     'discount', 'name', 'desc', 'type', 'short', 'in_stock']
]

# rename column 'product quantity' to 'qty' so it takes up less space
discounts = discounts.rename(columns={'product_quantity': 'qty'})

discounts
```

## Setting the Price Tier

```python
categories = final_df['Category'].unique()
categoryPriceRelationDfCheap = pd.DataFrame()
categoryPriceRelationDfExpensive = pd.DataFrame()
counter = 0
for cat in categories:
  quantileThreshhold = final_df.loc[final_df['Category']==cat, 'price'].quantile(.80)
  categoryPriceRelationDfExpensive.loc[counter, 'Category'] = cat
  categoryPriceRelationDfExpensive.loc[counter, 'price_category'] = 'expensive'
  categoryPriceRelationDfExpensive.loc[counter, 'price_threshhold'] = quantileThreshhold
  #cheap ones, because possible: (efficient code means not 2 unnessessary loops %D)
  quantileThreshhold = final_df.loc[final_df['Category']==cat, 'price'].quantile(.20)
  categoryPriceRelationDfCheap.loc[counter, 'Category'] = cat
  categoryPriceRelationDfCheap.loc[counter, 'price_category'] = 'cheap'
  categoryPriceRelationDfCheap.loc[counter, 'price_threshhold'] = quantileThreshhold
  counter=counter+1
  #counter2=counter2+2
categoryPriceRelationDfExpensive
```

```python
#category, price_category, price_threshhold - Table/DF Called... categoryPriceRelationDf
categories = final_df['Category'].unique()
categoryPriceRelationDf = pd.DataFrame()
counter = 0
counter2 = 1
for cat in categories:
  quantileThreshhold = final_df.loc[final_df['Category']==cat, 'price'].quantile(.80)
  categoryPriceRelationDf.loc[counter, 'Category'] = cat
  categoryPriceRelationDf.loc[counter, 'price_category'] = 'expensive'
  categoryPriceRelationDf.loc[counter, 'price_threshhold'] = quantileThreshhold
  #cheap ones, because possible: (efficient code means not 2 unnessessary loops %D)
  quantileThreshhold = final_df.loc[final_df['Category']==cat, 'price'].quantile(.20)
  categoryPriceRelationDf.loc[counter2, 'Category'] = cat
  categoryPriceRelationDf.loc[counter2, 'price_category'] = 'cheap'
  categoryPriceRelationDf.loc[counter2, 'price_threshhold'] = quantileThreshhold
  counter=counter+2
  counter2=counter2+2
categoryPriceRelationDf
```

```python
final_df = final_df.drop(
    columns=['cheap_threshhold', 'expensive_threshhold'],
    errors='ignore'
)
final_df = final_df.merge(categoryPriceRelationDfExpensive[['Category', 'price_threshhold']], on="Category", how='left').rename(columns={'price_threshhold': 'expensive_threshhold'})
final_df = final_df.merge(categoryPriceRelationDfCheap[['Category', 'price_threshhold']], on="Category", how='left').rename(columns={'price_threshhold': 'cheap_threshhold'})
final_df['priceTier']='medium'

final_df.loc[final_df['price']<final_df['cheap_threshhold'], 'priceTier'] = 'cheap'
final_df.loc[final_df['price']>final_df['expensive_threshhold'], 'priceTier'] = 'expensive'
```
